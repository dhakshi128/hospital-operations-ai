```python
# Bronze Layer — Raw Ingestion
df_raw = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("nullValue", "?") \
    .load("/Volumes/workspace/default/hospital_operations_ai/diabetic_data.csv")

print(f"Rows: {df_raw.count()}")
print(f"Columns: {len(df_raw.columns)}")
df_raw.printSchema()
```

```python
df_raw.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.bronze_hospital_raw")

print("Bronze table created successfully")
```
